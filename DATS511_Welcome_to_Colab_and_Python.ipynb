A variable gives a value a name. Python has integers (`int`), decimal numbers (`float`), text (`str`), and Boolean values (`bool`). Use `=` to assign a value and `==` to compare values. A line beginning with `#` is a comment.

## 4. Collections, loops, and decisions

A **list** stores several values. List indexing begins at zero. A `for` loop repeats an action, and `if` selects an action based on a condition. Indentation defines the body of a loop or condition.


```python
scores = [72, 85, 91, 68]
print("First score:", scores[0])
for score in scores:
    if score >= 70:
        print(f"{score}: passed")
    else:
        print(f"{score}: keep practicing")
average_score = sum(scores) / len(scores)
print(f"Average score: {average_score:.1f}")
```

## 5. Functions make code reusable

A function receives inputs and can return a result. Give functions clear names and explain their purpose. This function checks that the list contains at least one score before dividing.


```python
def mean_score(values):
    """Return the arithmetic mean of a nonempty collection of scores."""
    if not values:
        raise ValueError("Provide at least one score.")
    return sum(values) / len(values)

print("Mean:", mean_score(scores))
print("Another class:", mean_score([80, 90, 100]))
```

## 6. A small data example

Python can read tabular data. Here we use the built-in `csv` module, so no package installation or external file is required. CSV values start as text; convert them to numbers before calculating.


```python
import csv
from io import StringIO

sample_csv = "name,score\nAlex,72\nSam,85\nTaylor,91\nJordan,68\n"
rows = list(csv.DictReader(StringIO(sample_csv)))
for row in rows:
    row["score"] = int(row["score"])
print("Students:", len(rows))
print("Class average:", mean_score([row["score"] for row in rows]))
print("Highest score:", max(rows, key=lambda row: row["score"]))
```

## 9. Read errors and troubleshoot

An error message is feedback. Read its final line for the exception type and message, then inspect the line named in the traceback.

| Symptom | What to check |
|---|---|
| `NameError` | Spelling and whether the defining cell ran |
| `SyntaxError` | Quotes, parentheses, colons, and Python syntax |
| `IndentationError` | Consistent indentation inside blocks |
| `ModuleNotFoundError` | Correct package name and installation in this session |
| `FileNotFoundError` | Filename, path, and whether the file exists in this runtime |
| Unexpected result | Input types, formula, and execution order |
